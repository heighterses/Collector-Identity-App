# AI Provider Abstraction Layer

This notebook defines and tests an abstraction layer for communicating with an AI provider.
The goal is to ensure:
- Vendor independence
- Controlled prompt submission
- Testability using mock providers
- Clean separation between business logic and AI APIs


```python
from abc import ABC, abstractmethod
from dataclasses import dataclass
from typing import Optional
```

```python
@dataclass
class ArtworkContext:
    title: str
    artist: Optional[str]
    description: str
```

```python
class AIProvider(ABC):
    """
    Abstract base class for AI providers.
    Any AI provider must implement this interface.
    """

    @abstractmethod
    def generate_reflection(self, prompt: str) -> str:
        """
        Generate a reflection based on a prompt.
        """
        pass
```

```python
class MockAIProvider(AIProvider):
    """
    Mock AI provider used for local testing and development.
    """

    def generate_reflection(self, prompt: str) -> str:
        return (
            "This artwork seems to invite a quiet moment of reflection, "
            "allowing personal feelings to surface without demanding a fixed meaning."
        )
```

```python
mock_provider = MockAIProvider()

test_prompt = "Test prompt about an abstract painting."
output = mock_provider.generate_reflection(test_prompt)

print(output)
```

```python
class OpenAIProvider(AIProvider):
    """
    Concrete AI provider using an external LLM service.
    """

    def __init__(self, api_key: str, model_name: str):
        self.api_key = api_key
        self.model_name = model_name

    def generate_reflection(self, prompt: str) -> str:
        raise NotImplementedError(
            "Real API integration will be implemented after prompt design."
        )
```

```python
def get_reflection(provider: AIProvider, prompt: str) -> str:
    return provider.generate_reflection(prompt)


reflection = get_reflection(
    provider=mock_provider,
    prompt="A calm landscape painting at dusk."
)

print(reflection)
```

## Engineering Notes

- The AIProvider abstraction prevents vendor lock-in.
- MockAIProvider enables safe testing and development.
- No prompt logic or business rules exist in the provider layer.
- This layer will be reused unchanged by the backend service.
