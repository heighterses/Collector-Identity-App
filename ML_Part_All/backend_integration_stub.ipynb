# Backend Integration Stub for AI Reflection Generation

This notebook simulates how a backend service integrates with the AI
reflection generation pipeline.

It demonstrates:
- Data flow between backend and AI layer
- Controlled AI invocation
- User confirmation gating persistence
- Clear separation of concerns

This is a stub, not a production backend.


```python
from dataclasses import dataclass
from typing import Optional, Dict
```

```python
@dataclass
class ArtworkRecord:
    artwork_id: int
    title: str
    artist: Optional[str]
    description: str
    image_object_key: str  # MinIO reference


@dataclass
class ReflectionRecord:
    artwork_id: int
    reflection_text: str
    is_user_confirmed: bool
```

```python
DATABASE: Dict[str, list] = {
    "artworks": [],
    "reflections": []
}
```

```python
class MinIOClientStub:

    def get_object(self, object_key: str) -> str:
        return f"[Binary image data for {object_key}]"
```

```python
class AIProvider:

    def generate_reflection(self, prompt: str) -> str:
        return (
            "The artwork holds a gentle stillness, creating space for "
            "personal emotions to surface naturally and without pressure."
        )
```

```python
def build_prompt(title: str, description: str) -> str:
    return f"""
SYSTEM:
You are a reflective writing assistant.

USER:
Artwork title: {title}
Description: {description}

Write a gentle, non-authoritative reflection.
""".strip()
```

```python
def generate_reflection_service(
    artwork: ArtworkRecord,
    ai_provider: AIProvider
) -> str:

    prompt = build_prompt(
        title=artwork.title,
        description=artwork.description
    )

    return ai_provider.generate_reflection(prompt)
```

```python
def save_reflection(
    reflection: ReflectionRecord
):
    if not reflection.is_user_confirmed:
        raise ValueError("Reflection cannot be saved without user confirmation.")

    DATABASE["reflections"].append(reflection)
```

```python
# Create artwork record (already stored earlier)
artwork = ArtworkRecord(
    artwork_id=1,
    title="Evening Stillness",
    artist=None,
    description="Soft hills beneath a fading sky.",
    image_object_key="artworks/1/image.jpg"
)

DATABASE["artworks"].append(artwork)

# Initialize services
ai_provider = AIProvider()
minio_client = MinIOClientStub()

# Backend fetches image (not sent to AI)
image_data = minio_client.get_object(artwork.image_object_key)
print("Image retrieved from MinIO (not sent to AI).")

# Generate reflection
reflection_text = generate_reflection_service(
    artwork=artwork,
    ai_provider=ai_provider
)

print("\nGenerated Reflection:\n")
print(reflection_text)
```
